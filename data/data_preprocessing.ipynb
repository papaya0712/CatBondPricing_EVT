```python
import pandas as pd
import csv

csv_path_1 = r"C:\Users\nikla\Desktop\github_projects\CatBondPricing_EVT\data\earthquakes_us_1950-1987.feather"
csv_path_2 = r"C:\Users\nikla\Desktop\github_projects\CatBondPricing_EVT\data\earthquakes_us_1988-2010.feather"
raw_data_1 = pd.read_feather(csv_path_1)
raw_data_2 = pd.read_feather(csv_path_2)

raw_data = pd.concat([raw_data_1, raw_data_2], ignore_index=True)
raw_data

```

```python
data = raw_data
data = data[data["place"].str.endswith((", CA", ", California"), na=False)]
data = data.reset_index(drop=True)
data["time"] = pd.to_datetime(data["time"])

data = data.drop(columns=
                     ["type", "status", "updated", "magNst", 
                      "locationSource", "magSource", "depthError", 
                      "horizontalError", "nst", "gap", "place"]
                     )


output_path = r"C:\Users\nikla\Desktop\github_projects\CatBondPricing_EVT\data\earthquakes_california_1950-2010.feather"
data.to_feather(output_path)
```
